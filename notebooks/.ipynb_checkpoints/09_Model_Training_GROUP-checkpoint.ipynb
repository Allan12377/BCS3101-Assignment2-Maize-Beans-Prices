---
## 2. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

print("Libraries imported.")
```

---
## 3. Load the filtered data

We use the same seven-market file that Notebook 1 prepared.

```python
df = pd.read_csv('uganda_maize_beans_selected_markets.csv')

print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
print("\nMarkets:")
print(df['mkt_name'].value_counts())
```

---
## 4. Prepare features and targets

We repeat the same encoding and column choices used in the earlier notebooks so this file can run on its own.

```python
# One-hot encode market names (nominal variable → no natural order)
df = pd.get_dummies(df, columns=['mkt_name'], drop_first=True, dtype=float)
market_cols = [c for c in df.columns if c.startswith('mkt_name_')]

feature_candidates = [
    'year', 'month', 'lat', 'lon',
    'c_oil', 'c_food_price_index',
    'inflation_maize', 'inflation_beans',
    'trust_maize', 'trust_beans',
    'data_coverage', 'index_confidence_score'
] + market_cols

feature_cols = [c for c in feature_candidates if c in df.columns]

df_model = df[feature_cols + ['c_maize', 'c_beans']].dropna().copy()

print(f"Rows after dropping missing values: {df_model.shape[0]}")
print(f"Number of features: {len(feature_cols)}")
print("Features:", feature_cols)
```

---
## 5. Train / test split

We keep 80% of the rows for training and 20% for testing.  
A fixed random seed makes the split reproducible.

```python
rng = np.random.RandomState(42)
indices = rng.permutation(len(df_model))
split = int(0.8 * len(df_model))
train_idx = indices[:split]
test_idx  = indices[split:]

X = df_model[feature_cols].values.astype(float)
y_maize = df_model['c_maize'].values.astype(float)
y_beans = df_model['c_beans'].values.astype(float)

X_train, X_test = X[train_idx], X[test_idx]
ym_train, ym_test = y_maize[train_idx], y_maize[test_idx]
yb_train, yb_test = y_beans[train_idx], y_beans[test_idx]

print(f"Training rows: {len(train_idx)}")
print(f"Test rows:     {len(test_idx)}")
```

---
## 6. Scale the features

We fit the mean and standard deviation on the **training** data only.  
Then we transform both the training and the test sets.  
This follows the guide’s rule against data leakage.

```python
mu = X_train.mean(axis=0)
sigma = X_train.std(axis=0)
sigma[sigma == 0] = 1.0   # avoid division by zero

X_train_s = (X_train - mu) / sigma
X_test_s  = (X_test  - mu) / sigma

print("Features scaled using training-set mean and standard deviation.")
print(f"Training shape: {X_train_s.shape}")
print(f"Test shape:     {X_test_s.shape}")
```

---
## 7. Train linear regression (ordinary least squares)

We add a column of ones for the intercept and solve the normal equations with numpy.  
This is the classic linear regression formula:

β = (XᵀX)⁻¹ Xᵀy

```python
def add_intercept(A):
    """Add a column of ones so the model has an intercept."""
    return np.column_stack([np.ones(len(A)), A])

def fit_ols(Xtr, ytr):
    """Fit ordinary least squares and return the coefficient vector."""
    beta, residuals, rank, s = np.linalg.lstsq(Xtr, ytr, rcond=None)
    return beta

def predict_ols(Xte, beta):
    return Xte @ beta

Xt = add_intercept(X_train_s)
Xs = add_intercept(X_test_s)

beta_maize = fit_ols(Xt, ym_train)
beta_beans = fit_ols(Xt, yb_train)

pred_maize = predict_ols(Xs, beta_maize)
pred_beans = predict_ols(Xs, beta_beans)

print("Linear models fitted for maize and beans.")
print(f"Number of coefficients (including intercept): {len(beta_maize)}")
```

---
## 8. Evaluate the models

We report three common regression metrics:

- **MAE** (Mean Absolute Error) – average size of the error in UGX  
- **RMSE** (Root Mean Squared Error) – penalises large errors more  
- **R²** (coefficient of determination) – how much of the variance is explained (1.0 is perfect)

```python
def mae(actual, predicted):
    return np.mean(np.abs(actual - predicted))

def rmse(actual, predicted):
    return np.sqrt(np.mean((actual - predicted) ** 2))

def r_squared(actual, predicted):
    ss_res = np.sum((actual - predicted) ** 2)
    ss_tot = np.sum((actual - np.mean(actual)) ** 2)
    return 1 - ss_res / ss_tot if ss_tot > 0 else 0.0

print("=== Maize price model ===")
print(f"MAE:  {mae(ym_test, pred_maize):.1f} UGX")
print(f"RMSE: {rmse(ym_test, pred_maize):.1f} UGX")
print(f"R²:   {r_squared(ym_test, pred_maize):.3f}")

print("\n=== Beans price model ===")
print(f"MAE:  {mae(yb_test, pred_beans):.1f} UGX")
print(f"RMSE: {rmse(yb_test, pred_beans):.1f} UGX")
print(f"R²:   {r_squared(yb_test, pred_beans):.3f}")
```

**Short interpretation**  
An R² above 0.8 means the linear model explains a large part of the price variation on the test set.  
The remaining error (MAE and RMSE) shows the typical size of the mistakes in Uganda shillings.  
These numbers give a baseline. A more flexible model (for example a random forest) could be tried later if the group wants higher accuracy.

---
## 9. Visual check – predicted versus actual

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(ym_test, pred_maize, alpha=0.5, s=15, color='0.3')
lims = [min(ym_test.min(), pred_maize.min()), max(ym_test.max(), pred_maize.max())]
axes[0].plot(lims, lims, 'k--', linewidth=1)
axes[0].set_xlabel('Actual maize price (UGX)')
axes[0].set_ylabel('Predicted maize price (UGX)')
axes[0].set_title('Maize: predicted vs actual')

axes[1].scatter(yb_test, pred_beans, alpha=0.5, s=15, color='0.3')
lims = [min(yb_test.min(), pred_beans.min()), max(yb_test.max(), pred_beans.max())]
axes[1].plot(lims, lims, 'k--', linewidth=1)
axes[1].set_xlabel('Actual beans price (UGX)')
axes[1].set_ylabel('Predicted beans price (UGX)')
axes[1].set_title('Beans: predicted vs actual')

plt.tight_layout()
plt.show()
```

Points that lie close to the dashed line are good predictions.  
Points far from the line are larger errors.

---
## 10. Save the prediction results

We build a clear table that contains the actual price, the predicted price and the error for every test row.  
This table is saved as a CSV file so the group can open it in Excel or share it.

```python
meta = df_model.iloc[test_idx][['year', 'month']].reset_index(drop=True)

results = meta.copy()
results['actual_maize']    = np.round(ym_test, 1)
results['predicted_maize'] = np.round(pred_maize, 1)
results['error_maize']     = np.round(results['actual_maize'] - results['predicted_maize'], 1)
results['abs_error_maize'] = np.abs(results['error_maize'])

results['actual_beans']    = np.round(yb_test, 1)
results['predicted_beans'] = np.round(pred_beans, 1)
results['error_beans']     = np.round(results['actual_beans'] - results['predicted_beans'], 1)
results['abs_error_beans'] = np.abs(results['error_beans'])

results = results.sort_values(['year', 'month']).reset_index(drop=True)

print("First 8 rows of the results table:")
print(results.head(8).to_string())

results.to_csv('prediction_results_maize_beans.csv', index=False)
print("\nResults saved to prediction_results_maize_beans.csv")
```

---
## 11. Summary of absolute errors

```python
print(results[['abs_error_maize', 'abs_error_beans']].describe().round(1))
```
